```python
import json
import pandas as pd

# 1. Đọc dữ liệu JSON
input_path = "/content/drive/MyDrive/Temp_KLTN/MOOCCubeX/Dataset/entities/course.json"
output_path = "/content/drive/MyDrive/Temp_KLTN/resource_video_course.parquet"

with open(input_path, "r", encoding="utf-8") as f:
    courses = [json.loads(line) for line in f if line.strip()]

# 2. Xử lý dữ liệu
data = []

for course in courses:
    course_id = course.get("id")

    # Các danh sách lưu thông tin video của khóa học
    titles_chapter = []
    title_video = []
    chapters = []
    video_ids = []

    # Duyệt qua danh sách resource
    for resource in course.get("resource", []):

        resource_id = resource.get("resource_id", "")

        # Chỉ lấy resource là Video
        if not resource_id.startswith("V_"):
            continue

        titles = resource.get("titles") or []

        # Lấy thông tin tiêu đề
        chapter_title = titles[0] if len(titles) > 0 else None
        video_title = titles[2] if len(titles) > 2 else None

        titles_chapter.append(chapter_title)
        title_video.append(video_title)
        chapters.append(resource.get("chapter"))
        video_ids.append(resource_id)

    # 3. Lưu thông tin từng khóa học
    data.append({
        "id": course_id,
        "titles_chapter": titles_chapter,
        "title_video": title_video,
        "chapter": chapters,
        "video_id": video_ids,
        "video_counts": len(video_ids)
    })

# 4. Chuyển thành DataFrame
df = pd.DataFrame(data)

# 5. Xuất file Parquet
df.to_parquet(output_path, index=False, engine="pyarrow")

print("Hoàn thành xử lý dữ liệu!")
print(f"Số lượng khóa học: {len(df)}")
print(f"Tổng số video: {df['video_counts'].sum()}")
print(f"Đã lưu tại: {output_path}")

# 6. Kiểm tra dữ liệu
df.head()
df.info()
```
